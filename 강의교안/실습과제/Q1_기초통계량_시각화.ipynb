## 문제 2: 기술통계량 계산 (20점)

**Air temperature [K]** 컬럼의 기술통계량을 계산하세요.

## 문제 5: 그룹별 통계량 비교 (30점)

**Machine failure** (불량 여부)별로 온도 평균을 비교하세요.

```python
# 불량 여부별 건수
print("=== 불량 여부별 건수 ===")
print(df['Machine failure'].value_counts())
```

```python
# TODO: 불량 여부별 평균 온도 계산 (빈칸 채우기)
group_mean = df.groupby('______')['Air temperature [K]'].mean()  # 'Machine failure'로 그룹화
print("\n=== 불량 여부별 평균 온도 ===")
print(group_mean)
```

```python
# 막대 그래프로 시각화
group_mean.plot(kind='bar', color=['green', 'red'], edgecolor='black')
plt.xlabel('Machine Failure (0: Normal, 1: Failure)')
plt.ylabel('Average Air Temperature (K)')
plt.title('Average Temperature by Machine Failure')
plt.xticks(rotation=0)
plt.show()
```
