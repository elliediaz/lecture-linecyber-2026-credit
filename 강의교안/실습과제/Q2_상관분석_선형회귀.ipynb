```python
# 필요한 라이브러리 임포트
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# 데이터 로드
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00601/ai4i2020.csv"
df = pd.read_csv(url)
print(f"데이터 로드 완료: {df.shape}")
```

공정 온도와 회전 속도의 **피어슨 상관계수**를 계산하세요.

```python
# 두 변수 정의
x_col = 'Process temperature [K]'
y_col = 'Rotational speed [rpm]'

# TODO: 상관계수 계산 (빈칸 채우기)
corr_value = df[x_col].______(df[y_col])  # .corr() 메서드 사용
print(f"상관계수: {corr_value:.4f}")
```

## 문제 3: 선형회귀 모델 학습 (25점)

sklearn의 **LinearRegression**을 사용하여 예측 모델을 만드세요.

```python
# 특성(X)과 타겟(y) 정의
# 주의: X는 2차원 배열이어야 합니다 (대괄호 2개)
X = df[[x_col]]
y = df[y_col]

print(f"X shape: {X.shape}")
print(f"y shape: {y.shape}")
```

```python
# TODO: 모델 생성 및 학습 (빈칸 채우기)
model = LinearRegression()
model.______(X, y)  # fit 메서드 사용

# 회귀계수와 절편 출력
print(f"회귀계수 (기울기): {model.coef_[0]:.4f}")
print(f"절편: {model.intercept_:.4f}")
```

## 문제 4: 산점도에 회귀선 추가 (15점)

산점도 위에 회귀선을 그려서 모델의 적합도를 시각적으로 확인하세요.

```python
# 예측값 계산
y_pred = model.predict(X)

# 시각화
plt.figure(figsize=(10, 6))
plt.scatter(df[x_col], y, alpha=0.5, label='Actual')
plt.plot(df[x_col], y_pred, color='red', linewidth=2, label='Regression Line')
plt.xlabel('Process Temperature (K)')
plt.ylabel('Rotational Speed (rpm)')
plt.title('Linear Regression: Temperature vs Rotational Speed')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 문제 5: 모델 평가 및 예측 (20점)

**R² 점수**로 모델의 적합도를 평가하고, 새로운 값을 예측하세요.

```python
# TODO: R² 점수 계산 (빈칸 채우기)
r2 = r2_score(y, y_pred)
print(f"R² Score (r2_score 함수): {r2:.4f}")

# 또는 model.score() 사용
r2_model = model.______(X, y)  # score 메서드 사용
print(f"R² Score (model.score): {r2_model:.4f}")
```

```python
# 새로운 값 예측
new_temp = [[310.0]]  # 310K
predicted_speed = model.predict(new_temp)
print(f"\n310K에서의 예측 회전속도: {predicted_speed[0]:.2f} rpm")
```
